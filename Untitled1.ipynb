```python
# MNIST 데이터를 활용하여 80%의 훈련 데이터로 로지스틱 회귀 분류기를 구축한 뒤, 검증 데이터의 첫 25개를 사용하여 0부터 9까지의 숫자 식별 성능을 시각화하시오.

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# 1. MNIST 데이터셋 로드
print("MNIST 데이터를 로드 중입니다...")
(x_train_full, y_train_full), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# 데이터 전처리
# 픽셀 값을 0-1 범위로 정규화
x_train_full = x_train_full / 255.0
x_test = x_test / 255.0

# 로지스틱 회귀를 위해 이미지를 평탄화 (28x28 = 784 특징)
x_train_full_flat = x_train_full.reshape(-1, 28 * 28)
x_test_flat = x_test.reshape(-1, 28 * 28)

# 2. 훈련 및 검증 세트로 데이터 분할 (훈련용 80%)
# 전체 훈련 세트는 60,000개의 샘플을 가짐
validation_size = int(len(x_train_full_flat) * 0.2) # 검증용 20%

x_train, x_val = x_train_full_flat[validation_size:], x_train_full_flat[:validation_size]
y_train, y_val = y_train_full[validation_size:], y_train_full[:validation_size]

print(f"훈련 데이터 형태: {x_train.shape}, {y_train.shape}")
print(f"검증 데이터 형태: {x_val.shape}, {y_val.shape}")
print(f"테스트 데이터 형태: {x_test_flat.shape}, {y_test.shape}")

# 3. Keras를 사용하여 로지스틱 회귀 모델 구축
# 로지스틱 회귀 모델은 다중 클래스 분류를 위한 소프트맥스 활성화 함수를 가진 단일 계층 신경망입니다.
model = tf.keras.models.Sequential([
    tf.keras.layers.Dense(10, activation='softmax', input_shape=(784,))
])

# 모델 컴파일
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy', # 정수 레이블에 sparse_categorical_crossentropy 사용
    metrics=['accuracy']
)

# 모델 요약 출력
model.summary()

# 4. 모델 훈련
print("로지스틱 회귀 모델을 훈련 중입니다...")
history = model.fit(
    x_train, y_train,
    epochs=10, # 에포크 수 조정 가능
    validation_data=(x_val, y_val),
    verbose=0 # 진행률 바를 위해 1로 설정
)

print("모델 훈련 완료.")

# 5. 테스트 세트에서 모델 평가
loss, accuracy = model.evaluate(x_test_flat, y_test, verbose=0)
print(f"테스트 정확도: {accuracy:.4f}")
print(f"테스트 손실: {loss:.4f}")

# 6. 첫 25개 검증 데이터 포인트를 사용하여 성능 시각화
print("시각화를 위한 예측을 생성 중입니다...")
predictions = model.predict(x_val[:25])
predicted_labels = np.argmax(predictions, axis=1)

plt.figure(figsize=(10, 10))
for i in range(25):
    plt.subplot(5, 5, i + 1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(x_val[i].reshape(28, 28), cmap=plt.cm.binary)
    true_label = y_val[i]
    pred_label = predicted_labels[i]
    color = 'green' if true_label == pred_label else 'red'
    plt.xlabel(f"예측: {pred_label}\n실제: {true_label}", color=color)
plt.suptitle("첫 25개 검증 샘플: 예측 대 실제 레이블", fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
